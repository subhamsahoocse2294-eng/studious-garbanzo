```python
# 1. Sum and average of first 55 natural numbers
print("=" * 50)
print("1. Sum and Average of first 55 natural numbers")
print("=" * 50)

n = 55
total = 0
for i in range(1, n + 1):
    total += i

average = total / n
print(f"Sum of first {n} natural numbers = {total}")
print(f"Average of first {n} natural numbers = {average}")


# 2. Calculating Factorial of 7 with a Loop
print("\n" + "=" * 50)
print("2. Factorial of 7 using a Loop")
print("=" * 50)

num = 7
factorial = 1
for i in range(1, num + 1):
    factorial *= i

print(f"Factorial of {num} = {factorial}")


# 3. Reverse a string
print("\n" + "=" * 50)
print("3. Reverse a String")
print("=" * 50)

text = input("Enter a string to reverse (e.g. remember): ")
reversed_text = ""
for char in text:
    reversed_text = char + reversed_text

print(f"Original string : {text}")
print(f"Reversed string : {reversed_text}")


# 4. Check whether a year is leap year or not
print("\n" + "=" * 50)
print("4. Leap Year Checker")
print("=" * 50)

year = int(input("Enter a year: "))

if (year % 4 == 0 and year % 100 != 0) or (year % 400 == 0):
    print(f"{year} is a Leap Year")
else:
    print(f"{year} is NOT a Leap Year")
```
